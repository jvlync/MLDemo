```python
# @title Import pandas library
import pandas as pd
```

```python
# @title Create and print DataFrame

# Manually create DataFrame
example_df = pd.DataFrame({
    "Name": ["Mark", "Jane", "Bob", "Maria", "Alice", "Seth"],
    "Age": [20, 22, 21, 20, 19, 17],
    "Sex": ["M", "F", "M", "F", "F", "M"]
})

# Print DataFrame content
example_df
```

```python
# @title Create and print Series

# Manually create Series
example_series = pd.Series(["Mark", "Jane", "Bob", "Maria", "Alice", "Seth"], name="Name")

# Print Series content
example_series
```

```python
# @title A DataFrame is a combination of multiple Series

# Manually create Series
series_1 = pd.Series(["Mark", "Jane", "Bob", "Maria", "Alice", "Seth"])
series_2 = pd.Series([20, 22, 21, 20, 19, 17])
series_3 = pd.Series(["M", "F", "M", "F", "F", "M"])

# Manually create DataFrame from the Series
df = pd.DataFrame({"Name": series_1, "Age": series_2, "Sex":  series_3})

# Print DataFrame content
df
```

```python
# @title Scalar Operations

# Manually create Series
age_series = pd.Series([20, 22, 21, 20, 19, 17], name="Age")

# Do scalar operation and store results in another variable
results_series = age_series > 20

# Print Series content
results_series
```

```python
# @title Series to Series Operations

# Manually create Series
dist_series = pd.Series([10, 20, 30, 10, 100, 200, 300], name="Distance")
time_series = pd.Series([20, 10, 30, 100, 10], name="Time")

# Do series to series operation and store results in another variable
results_series = dist_series / time_series

# Print Series content
results_series
```
